**O que é Python?**

Python é uma linguagem de programação bastante difundida. Foi criada por Guido van Rossum e lançada no ano de 1991.

Aplicações da linguagem

Python é utilizada para:
* desenvolvimento web (no lado do servidor);

* desenvolvimento de software;

* cálculos matemáticos;

* automação de tarefas do sistema.

**Por que escolher Python?**

Python é compatível com diferentes plataformas (Windows, macOS, Linux, Raspberry Pi, entre outras).

Sua sintaxe é simples e assemelha-se à língua inglesa.
A estrutura sintática do Python permite que programadores escrevam programas com menos linhas se comparado a outras linguagens de programação.

Python opera por meio de um sistema interpretador, o que significa que o código pode ser executado tão logo seja escrito. Esse recurso torna a prototipagem bastante ágil.

Python pode ser utilizado segundo diferentes paradigmas de programação: procedural, orientado a objetos ou funcional.





**Sintaxe do Python em comparação com outras linguagens de programação**

Python foi concebido tendo em vista a legibilidade, apresentando semelhanças com a língua inglesa e recebendo influências da linguagem matemática.

Python utiliza mudanças de linha para delimitar comandos, diferentemente de outras linguagens de programação que frequentemente empregam ponto e vírgula ou parênteses para essa finalidade.

Python recorre à indentação — por meio de espaços em branco — para definir escopos, tais como os de laços de repetição, funções e classes. Outras linguagens de programação, por sua vez, comumente utilizam chaves para essa mesma finalidade.
